```python
import pandas as pd
import numpy as np
import math

def format_sig_str(x, sig=17):
    """Return a string with <= sig significant digits (no scientific notation)."""
    if pd.isna(x):
        return ""
    try:
        x = float(x)
    except:
        return x

    # Convert to scientific, truncate significant digits manually
    s = f"{x:.17g}"

    # Remove scientific notation entirely
    if "e" in s or "E" in s:
        s = f"{x:.17f}".rstrip("0").rstrip(".")

    # Ensure no more than sig significant digits
    digits = [c for c in s if c.isdigit()]
    if len(digits) > sig:
        # Truncate extra digits
        excess = len(digits) - sig
        count = 0
        out = []
        for c in s:
            if c.isdigit():
                if count < sig:
                    out.append(c)
                    count += 1
                else:
                    # drop remaining digits
                    continue
            else:
                out.append(c)
        s = "".join(out)

    return s

df = pd.read_csv("runs/RF/default_run_name/event_part_3.csv")

# Apply strict string formatting to all numeric columns
for col in df.columns:
    if np.issubdtype(df[col].dtype, np.number):
        df[col] = df[col].apply(format_sig_str)

# Force pandas NOT to convert back to float
df.to_csv("runs/RF/default_run_name/event_part_3_trim.csv", index=False)
```
